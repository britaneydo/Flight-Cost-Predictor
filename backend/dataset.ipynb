```python
df = pd.read_parquet("data/raw/lax_flightsLARGE.parquet")

# see which destinations have the most data, dataset is still too large to upload to gh, so we have to cut it down
df.destinationAirport.value_counts()
```

```python
# top 3 airports!
df = df[df.destinationAirport.isin(["BOS", "ATL", "LGA"])]
# len of dataset to check if still too large
print(len(df))
```

```python
# new parquet file since old one was too large
df.to_parquet("data/lax_flights.parquet", index=False)

# new size is 81mb. you can check this by going to file manager -> right clicking on this file -> properties -> size
# for reference, before cleaning the dataset, it was 352mb. you can verify this by checking the raw folder in backend/data. this folder is in .gitignore, so it doesnt get committed to gh. gh only allows uploads of 100mb or less.
```
